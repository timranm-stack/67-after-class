```python
import matplotlib.pyplot as plt
```

```python
weeks = ["Week 1", "Week 2", "Week 3", "Week 4", "Week 5", "Week 6"]
savings = [25, 40, 35, 60, 80, 95]
```

```python
# 1. Create a customised line graph
plt.figure(figsize=(8, 4))
plt.plot(
    weeks,
    savings,
    color="green",      # Line colour
    marker="o",         # Marker style
    linestyle="--",     # Line style (dashed)
    linewidth=2         # Line width
)

# Add title and axis labels
plt.title("My Weekly Savings Progress (Line Chart)")
plt.xlabel("Weeks")
plt.ylabel("Savings ($)")
plt.grid(True)
plt.show()
```

```python
# 2. Create a bar chart to compare weekly savings
plt.figure(figsize=(8, 4))
plt.bar(weeks, savings, color="skyblue", edgecolor="navy")

# Add title and axis labels
plt.title("Weekly Savings Comparison (Bar Chart)")
plt.xlabel("Weeks")
plt.ylabel("Savings ($)")
plt.show()
```
