```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Set a clean theme for all visualizations
sns.set_theme(style="whitegrid")

# Load the built-in Seaborn 'tips' dataset
tips = sns.load_dataset("tips")
```

```python
# ---------------------------------------------------------
# 1. Bar Plot & Count Plot: Customer bill data and counts
# ---------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Bar plot: Average total bill by day and gender
sns.barplot(data=tips, x="day", y="total_bill", hue="sex", palette="pastel", ax=axes)
axes.set_title("Average Total Bill by Day and Gender")
axes.set_xlabel("Day of the Week")
axes.set_ylabel("Average Total Bill ($)")

# Count plot: Number of customers/visits by day and gender
sns.countplot(data=tips, x="day", hue="sex", palette="pastel", ax=axes)
axes.set_title("Customer Visit Count by Day and Gender")
axes.set_xlabel("Day of the Week")
axes.set_ylabel("Number of Visits")

plt.tight_layout()
plt.show()
```

```python
# ---------------------------------------------------------
# 2. Box Plot, Strip Plot & Swarm Plot: Spread of bill amounts
# ---------------------------------------------------------
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Box plot
sns.boxplot(data=tips, x="day", y="total_bill", palette="Set2", ax=axes)
axes.set_title("Box Plot: Total Bill Spread by Day")
axes.set_xlabel("Day of the Week")
axes.set_ylabel("Total Bill ($)")

# Strip plot
sns.stripplot(data=tips, x="day", y="total_bill", jitter=True, palette="Set2", ax=axes)
axes.set_title("Strip Plot: Individual Bills by Day")
axes.set_xlabel("Day of the Week")
axes.set_ylabel("Total Bill ($)")

# Swarm plot
sns.swarmplot(data=tips, x="day", y="total_bill", palette="Set2", ax=axes)
axes.set_title("Swarm Plot: Non-Overlapping Bills by Day")
axes.set_xlabel("Day of the Week")
axes.set_ylabel("Total Bill ($)")

plt.tight_layout()
plt.show()
```

```python
# ---------------------------------------------------------
# 3. Joint Plot & Pair Plot: Relationships between numeric values
# ---------------------------------------------------------
# Joint plot: Total bill vs. tip amount
joint = sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="scatter",
    hue="time",
    palette="coolwarm"
)
joint.fig.suptitle("Joint Plot: Total Bill vs. Tip Amount by Meal Time", y=1.02)
joint.set_axis_labels("Total Bill ($)", "Tip Amount ($)")
plt.show()

# Pair plot: All numeric columns colored by gender
pair = sns.pairplot(tips, hue="sex", palette="husl")
pair.fig.suptitle("Pair Plot: Numeric Relationships in Tips Dataset", y=1.02)
plt.show()
```

```python
# ---------------------------------------------------------
# 4. Point Plot & lmplot: Compare averages and trend lines
# ---------------------------------------------------------
# Point plot: Average tip across party sizes by meal time
plt.figure(figsize=(9, 5))
sns.pointplot(data=tips, x="size", y="tip", hue="time", palette="dark")
plt.title("Point Plot: Average Tip by Party Size and Meal Time")
plt.xlabel("Party Size (Guests)")
plt.ylabel("Average Tip ($)")
plt.tight_layout()
plt.show()

# lmplot: Linear regression trend line between total bill and tip
lm = sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    height=5,
    aspect=1.4,
    palette="Set1"
)
lm.fig.suptitle("Regression Trend Line: Total Bill vs. Tip by Smoking Status", y=1.02)
lm.set_axis_labels("Total Bill ($)", "Tip Amount ($)")
plt.show()
```
