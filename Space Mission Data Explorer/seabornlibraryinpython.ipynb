```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Set a clean visual theme for all plots
sns.set_theme(style="whitegrid")
```

```python
# 1. Load and inspect the built-in Seaborn 'planets' dataset
planets = sns.load_dataset("planets")

print("--- First 5 Rows ---")
print(planets.head())

print("\n--- Dataset Info ---")
print(planets.info())

print("\n--- Summary Statistics ---")
print(planets.describe())
```

```python
# 2. Histogram: Distribution of planet orbital periods
plt.figure(figsize=(10, 5))
sns.histplot(data=planets, x="orbital_period", bins=40, color="steelblue")
plt.xscale("log")  # Log scale handles extreme orbital period outliers clearly
plt.title("Distribution of Planet Orbital Periods (Log Scale)")
plt.xlabel("Orbital Period (Days)")
plt.ylabel("Number of Planets")
plt.tight_layout()
plt.show()
```

```python
# 3. KDE Plot & Histogram with KDE Curve: Explore planet distance data
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Standalone KDE plot
sns.kdeplot(data=planets, x="distance", fill=True, color="teal", ax=axes[0])
axes[0].set_title("KDE Plot of Planet Distance")
axes[0].set_xlabel("Distance (Light Years)")
axes[0].set_ylabel("Density")

# Histogram with overlaid KDE curve
sns.histplot(data=planets, x="distance", bins=30, kde=True, color="coral", ax=axes[1])
axes[1].set_title("Histogram with KDE Curve of Planet Distance")
axes[1].set_xlabel("Distance (Light Years)")
axes[1].set_ylabel("Count")

plt.tight_layout()
plt.show()
```

```python
# 4. Scatter Plot: Compare distance and orbital period by discovery method
plt.figure(figsize=(11, 6))
sns.scatterplot(
    data=planets,
    x="distance",
    y="orbital_period",
    hue="method",
    alpha=0.75,
    palette="tab10"
)
plt.xscale("log")
plt.yscale("log")
plt.title("Planet Distance vs. Orbital Period by Discovery Method")
plt.xlabel("Distance (Light Years - Log Scale)")
plt.ylabel("Orbital Period (Days - Log Scale)")
plt.legend(title="Discovery Method", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# 5. Heatmap: Correlations between numeric planet data columns
numeric_planets = planets.select_dtypes(include="number")
correlation_matrix = numeric_planets.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)
plt.title("Correlation Heatmap of Numeric Planet Data")
plt.tight_layout()
plt.show()
```
